```python
%sql
CREATE OR REPLACE TABLE dbw_maritime_2026.maritime_showcase.gold_daily_port_traffic
USING DELTA AS
SELECT 
  CAST(event_time_utc AS DATE) AS report_date,
  COALESCE(current_port_vicinity, 'Open Sea') AS location_zone,
  COUNT(DISTINCT mmsi) AS unique_ships_count,
  COUNT(mmsi) AS total_signals_received
FROM dbw_maritime_2026.maritime_showcase.silver_ais_enriched
WHERE event_time_utc >= current_date() - INTERVAL 7 DAYS
GROUP BY CAST(event_time_utc AS DATE), COALESCE(current_port_vicinity, 'Open Sea')
ORDER BY report_date DESC, unique_ships_count DESC;
```
